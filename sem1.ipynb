```python
from numbers import Number

def get_min(input_list: list) -> Number:
    
    assert isinstance(input_list, list), "Входной параметр не является списком"
    assert len(input_list) > 0, "Входной параметр является пустым списком"

    current_min = input_list[0]
    for i in input_list: 
        assert isinstance(i, (int, float, complex)) and not isinstance(i, bool), "Входной параметр содержит не только числовые данные"
        if i < current_min: 
            current_min = i 

    return current_min

ind = []

example_list = [-5,-4,4556,123,44,-5,0,324]
min_value = get_min(example_list)

d = min_value

for i, j in enumerate(example_list):
    if j == d:
        ind.append(i)

print(ind)
```

```python
from random import randint


def puzirek(input_list: list):
   
    for i in range(len(input_list)-1):
        for j in range (len(input_list)-1-i):
            if input_list[j] > input_list[j+1]:
                input_list[j], input_list[j+1] = input_list[j+1], input_list[j]
    
    return input_list

massiv = [randint(0,1000) for i in range(10)]
test_m = puzirek(massiv)
print(test_m)
```

```python
import random

def nucleotide_generator(length):
    
    alphabet = ['A', 'G', 'T', 'U', 'C']
    
    for _ in range(length):
        yield random.choice(alphabet)

user_length = int(input())
gen = nucleotide_generator(user_length)
result_string = "".join(gen)
print(result_string)
```

```python
import random

def rand(input_int, answer, k):
    if input_int == answer:
        return print("Congratulation, u win in", k, "attempts")
    elif input_int > answer:
        print("Wrong, our number is lower, than ur. Attempts left:", 10-k)
        if k == 10:
            return print("AhAHhAhA, u looose")
    elif input_int < answer:
        print("Wrong, our number is higher, than ur. Attempts left:", 10-k)
        if k == 10:
            return print("AhAHhAhA, u looose")

a = random.randint(1,100)
k = 0
for i in range(10):
    k+=1
    b = int(input("Input ur number:"))
    rand(b, a, k)
```

```python
import random

def guess_the_number():
    """
    Запускает игру 'Угадай число' в диапазоне от 1 до 100 с 10 попытками.
    """
    secret_number = random.randint(1, 100)
    max_attempts = 10
    
    print("=== Игра 'Угадай число' ===")
    print(f"Я загадал число от 1 до 100. У вас есть {max_attempts} попыток.")
    print("-" * 30)

    for attempt in range(1, max_attempts + 1):
        try:
            # Запрашиваем ввод и пытаемся преобразовать в целое число
            user_input = input(f"[Попытка {attempt} из {max_attempts}] Введите число: ")
            user_guess = int(user_input)
        except ValueError:
            print("Ошибка: Пожалуйста, вводите только целые числа.")
            continue # Пропускаем итерацию, попытка сгорает (или можно не сжигать, если изменить логику)

        # Логика сравнения
        if user_guess == secret_number:
            print(f"\n🎉 Поздравляю! Вы угадали число {secret_number} с {attempt}-й попытки!")
            return  # Выход из функции при победе
        elif secret_number > user_guess:
            print("-> Загаданное число БОЛЬШЕ вашего.")
        else:
            print("-> Загаданное число МЕНЬШЕ вашего.")

    # Если цикл закончился, а return не сработал — значит попытки исчерпаны
    print("-" * 30)
    print(f"😔 К сожалению, попытки закончились. Было загадано число: {secret_number}")

# --- Запуск игры ---
if __name__ == "__main__":
    guess_the_number()
```
