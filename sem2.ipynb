### Computing GC Content

```python

#function of counting gc
def gc(seq):

    #creating the variable of GC-content
    ct_gc = 0 #var to counting GC
    ct_all = 0 #var to count all amk
    gc_main = -1.0 #most biggest %
    gc_work = 0.0 #current %
    name = "" #name of sequance of gc_main
    name_t = "" #timly name of current sequance

    for lines in seq: #get element
        lines = lines.strip() #remove spaces
        if lines.startswith('>'): #if section start with ">"
            if ct_all > 0: 
                gc_work = (ct_gc / ct_all) * 100 #counting procent
                
                #get the most biggest:
                if gc_work > gc_main:
                    gc_main = gc_work
                    name = name_t

            #ready to new section:
            name_t = lines.strip('>')
            ct_gc, ct_all = 0, 0
        else: #counting ATGC
            ct_gc += lines.count('G') + lines.count('C')
            ct_all += len(lines)

    #check last section, bcs it's skiped in loop
    if ct_all > 0:
                gc_work = (ct_gc / ct_all) * 100
                if gc_work > gc_main:
                    gc_main = gc_work
                    name = name_t
                
                name_t = lines.strip('>')
                ct_gc, ct_all = 0, 0

    #writing final result:
    res = [name, gc_main]
    return res

#open file
with open("rosalind_gc.txt") as file:
    seqs = file.readlines()

g = gc(seqs) #runnig function of counting

print(g)
```

### Counting Point Mutations

```python
#def function
def ham_ct(dna):
    res = 0 #final result
    str_1 = dna[0].strip() #get string 1
    str_2 = dna[1].strip() #get string 2

    #comprison:
    for i in range(len(str_1)):
        if str_1[i] != str_2[i]:
            res += 1
    return res


#open file
with open("rosalind_hamm.txt") as file:
    seqs = file.readlines()

g = ham_ct(seqs) #runnig function of counting

print(g)
```

### Majority Element

```python
file = open("rosalind_maj.txt", 'r')
data = list(map(int, file.read().split()))

k = data[0]
n = data[1]
result = []

for i in range(k):
        s = 2 + i * n
        a = data[s:s + n]
    
   
        found = -1
        for num in set(a):  # set для уникальных чисел
            if a.count(num) > n // 2:
                found = num
                break
    
        result.append(found)
        
print(' '.join(map(str, result)))
```

### Merge Two Sorted Arrays

```python
f = open('rosalind_mer.txt', 'r')

def puzirek(input_list: list):
   
    for i in range(len(input_list)-1):
        for j in range (len(input_list)-1-i):
            if input_list[j] > input_list[j+1]:
                input_list[j], input_list[j+1] = input_list[j+1], input_list[j]
    
    return input_list

data = list(map(int, f.read().split()))


n = data[0]
A = data[1:1+n]
m = data[1+n]
B = data[2+n:2+n+m]

C = A + B

C = puzirek(C)

print(' '.join(map(str, C)))
```

### Double-Degree Array

```python
f = open('rosalind_ddeg.txt', 'r')

data = f.read().split()


n = int(data[0])
m = int(data[1])


adj = [[] for i in range(n + 1)]  


for j in range(m):
    u = int(data[2 + j*2])
    v = int(data[2 + j*2 + 1])
    adj[u].append(v)
    adj[v].append(u)

# Вычисляем степени вершин
d = [0] * (n + 1)
for r in range(1, n + 1):
    d[r] = len(adj[r])


# Вычисляем сумму степеней соседей для каждой вершины
result = [0] * (n + 1)

for r in range(1, n + 1):
    summa = 0
    for neighbor in adj[r]:
        summa += d[neighbor]
    result[r] = summa


print(' '.join(map(str, result[1:n+1])))
```
