```python
class Rabbit:
    def __eats__(self):
        return f'clower'

class Beer:
    def __eats__(self):
        return f'berries'

class Octopus:
    def __eats__(self):
        return f'campers'
    
bear = Beer()
rab = Rabbit()
oct = Octopus()
print(bear.__eats__(), rab.__eats__(), oct.__eats__())
```

```python
class Laser:
    def __does__(self):
        return f'disintegrate'

class Claw:
    def __does__(self):
        return f'crush'

class SmartPhone:
    def __does__(self):
        return f'ring'

class Robot:
    def __does__(self):
        bear = Laser()
        rab = Claw()
        oct = SmartPhone()
        return f'robot can: {bear.__does__()}, {rab.__does__()}, {oct.__does__()}'

rob = Robot()
print(rob.__does__())
```
